# 2. Feature engineering
Create one leakage-aware customer row from the validated transaction extract. Features are interpretable and retain signed and absolute monetary behavior.

```python
import os, sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parents[1] / 'src'))
from transaction_analysis import load_extract, build_customer_features

transactions = load_extract(os.environ['DATA_PATH'])
as_of = transactions['transaction_date'].max().date()
customer_features = build_customer_features(transactions, as_of=as_of)
customer_features.head()
```

```python
feature_columns = [
    'transaction_count', 'gross_amount', 'net_amount', 'mean_transaction_amount',
    'debit_share', 'credit_share', 'transaction_type_count',
    'transaction_subtype_count', 'destination_bank_count', 'recency_days',
    'transactions_per_active_day'
]
customer_features.select(['customer_id', *feature_columns]).write_parquet('../customer_features.parquet')
customer_features.select(feature_columns).describe()
```

The feature table is the contract between extraction and modeling. Add domain-specific code mappings (for example type 51/subtype 12 = gas) as a separate lookup and aggregate category shares when those mappings are available.
