# 4. Evaluation and reporting
Check model stability and produce interactive customer segment and anomaly summaries for business review.

```python
import polars as pl
import plotly.express as px
segments = pl.read_parquet('../customer_segments.parquet')
anomalies = pl.read_parquet('../customer_anomalies.parquet')
segment_counts = segments.group_by('segment').len().sort('segment')
px.bar(segment_counts.to_pandas(), x='segment', y='len', title='Customers per segment').show()
anomalies.head(20)
```

## Acceptance checks
- No segment is too small to action without an explicit business reason.
- Refit with several random seeds and adjacent time windows; assignments and profiles should be materially stable.
- Compare anomaly candidates with known events and sample false positives.
- Export segment definitions, feature definitions, extraction window, model parameters, and review decisions alongside the Parquet outputs.

Recommended actions include tailored service journeys for recurring bill payers, cash-flow support for salary/expense patterns, and a review queue for high-confidence outliers.
