# 03 - Feature engineering

Objetivo: construir una matriz interpretable por cliente con frecuencia, montos, movimientos, diversidad y concentracion.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from analysis.pipeline import load_transactions, build_customer_features, save_table

ROOT = Path.cwd().parent if Path.cwd().name == "analysis" else Path.cwd()
tx = load_transactions(ROOT / "transacciones_sinteticas.csv")
customer_features = build_customer_features(tx)
customer_features.head()
```

```python
feature_groups = {
    "frequency": ["transaction_count", "transaction_type_count", "subtype_count"],
    "monetary": ["debit_total", "credit_total", "gross_amount", "net_amount", "amount_mean", "amount_median", "amount_std", "amount_iqr"],
    "behavior": ["debit_ratio", "credit_ratio", "bank_count"],
    "diversity": [c for c in customer_features.columns if "hhi" in c or "entropy" in c],
}
{group: [c for c in columns if c in customer_features.columns] for group, columns in feature_groups.items()}
```

```python
output = ROOT / "data" / "processed" / "customer_features.csv"
save_table(customer_features, output)
print(f"Guardado: {output} ({customer_features.height} clientes, {customer_features.width} columnas)")
```

## Indicadores
HHI alto representa concentracion; entropia alta representa diversidad. Las versiones normalizadas permiten comparar clientes sin interpretar los codigos como ordinales.
